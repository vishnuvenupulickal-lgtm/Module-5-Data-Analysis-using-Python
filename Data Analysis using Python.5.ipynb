```python
post_type_perf = (
    df.groupby("post_type", as_index=False)
    .agg(
        avg_engagement_rate=("engagement_rate", "mean"),
        avg_engagement_score=("engagement_score", "mean"),
        avg_likes=("likes", "mean"),
        post_count=("post_type", "count"),
    )
    .sort_values(by="avg_engagement_rate", ascending=False)
)
print(post_type_perf)
print("\nHighest engagement post type:", post_type_perf.iloc[0]["post_type"])
```

```python
category_perf = (
    df.groupby("post_category", as_index=False)
    .agg(
        avg_engagement_rate=("engagement_rate", "mean"),
        avg_engagement_score=("engagement_score", "mean"),
        avg_likes=("likes", "mean"),
        post_count=("post_category", "count"),
    )
    .sort_values(by="avg_engagement_rate", ascending=False)
)
print(category_perf)
print("\nBest-performing category:", category_perf.iloc[0]["post_category"])
```

```python
country_col = next(
    (c for c in df.columns if "country" in c.lower() or "region" in c.lower()),
    None,
)

if country_col:
    country_perf = (
        df.groupby(country_col, as_index=False)
        .agg(
            avg_engagement_rate=("engagement_rate", "mean"),
            avg_engagement_score=("engagement_score", "mean"),
            post_count=(country_col, "count"),
        )
        .sort_values(by="avg_engagement_rate", ascending=False)
    )
    print(country_perf.head(10))
    print("\nTop country:", country_perf.iloc[0][country_col])
else:
    print("No country column found in df. Available columns:", list(df.columns))
```

```python


age_col = next((c for c in df.columns if "age" in c.lower()), None)

if age_col:
    print("Correlation with engagement_rate:", df[age_col].corr(df["engagement_rate"]))
    print("Correlation with engagement_score:", df[age_col].corr(df["engagement_score"]))

    bins = [0, 17, 24, 34, 44, 54, 120]
    labels = ["<18", "18-24", "25-34", "35-44", "45-54", "55+"]
    age_groups = (
        df.assign(age_group=pd.cut(df[age_col], bins=bins, labels=labels))
        .groupby("age_group", observed=False, as_index=False)
        .agg(
            avg_engagement_rate=("engagement_rate", "mean"),
            avg_engagement_score=("engagement_score", "mean"),
            count=(age_col, "count"),
        )
    )
    print("\nEngagement by age group:")
    print(age_groups)
else:
    print("No age column found in df. Available columns:", list(df.columns))
```

```python
ver_col = next((c for c in df.columns if "verif" in c.lower()), None)

if ver_col:
    ver_perf = df.groupby(ver_col, as_index=False).agg(
        avg_engagement_rate=("engagement_rate", "mean"),
        avg_engagement_score=("engagement_score", "mean"),
        avg_likes=("likes", "mean"),
        avg_impressions=("impression_count", "mean"),
        count=(ver_col, "count"),
    )
    print(ver_perf)
else:
    print("No verified account column found in df. Available columns:", list(df.columns))
```

```python


df["posted_at"] = pd.to_datetime(df["posted_at"])
df["hour"] = df["posted_at"].dt.hour

# Average impressions by exact hour (0-23)
hourly_impressions = (
    df.groupby("hour", as_index=False)
    .agg(
        avg_impressions=("impression_count", "mean"),
        total_impressions=("impression_count", "sum"),
        post_count=("hour", "count"),
    )
    .sort_values(by="avg_impressions", ascending=False)
)
print("Top Hours by Average Impressions:")
print(hourly_impressions.head(5))
print(f"\nBest single hour: {int(hourly_impressions.iloc[0]['hour'])}:00")


# Average impressions by time of day period
def get_time_of_day(h):
    if 5 <= h < 12:
        return "Morning (05:00-11:59)"
    elif 12 <= h < 17:
        return "Afternoon (12:00-16:59)"
    elif 17 <= h < 21:
        return "Evening (17:00-20:59)"
    else:
        return "Night (21:00-04:59)"


df["time_of_day"] = df["hour"].apply(get_time_of_day)
tod_impressions = (
    df.groupby("time_of_day", as_index=False)["impression_count"]
    .mean()
    .sort_values(by="impression_count", ascending=False)
)
print("\nAverage Impressions by Time of Day:")
print(tod_impressions)
```

```python
watch_col = next(
    (
        c
        for c in df.columns
        if "watch" in c.lower() or "duration" in c.lower() or "time" in c.lower()
    ),
    None,
)
# Exclude 'posted_at' and 'time_of_day' if matched
if watch_col in ["posted_at", "time_of_day"]:
    watch_col = next(
        (
            c
            for c in df.columns
            if "watch" in c.lower() or "duration" in c.lower()
        ),
        None,
    )

if watch_col:
    device_watch = (
        df.groupby("device_type", as_index=False)
        .agg(
            avg_watch_time=(watch_col, "mean"),
            median_watch_time=(watch_col, "median"),
            avg_engagement_rate=("engagement_rate", "mean"),
            post_count=("device_type", "count"),
        )
        .sort_values(by="avg_watch_time", ascending=False)
    )
    print(device_watch)
else:
    print(
        "No watch_time column found. Comparing engagement metrics by device_type:"
    )
    device_eng = (
        df.groupby("device_type", as_index=False)
        .agg(
            avg_engagement_score=("engagement_score", "mean"),
            avg_engagement_rate=("engagement_rate", "mean"),
            avg_impressions=("impression_count", "mean"),
            avg_likes=("likes", "mean"),
            post_count=("device_type", "count"),
        )
        .sort_values(by="avg_engagement_score", ascending=False)
    )
    print(device_eng)
```

```python
sentiment_perf = (
    df.groupby("sentiment", as_index=False)
    .agg(
        post_count=("sentiment", "count"),
        avg_engagement_rate=("engagement_rate", "mean"),
        avg_engagement_score=("engagement_score", "mean"),
        avg_likes=("likes", "mean"),
        avg_impressions=("impression_count", "mean"),
    )
    .sort_values(by="avg_engagement_rate", ascending=False)
)
print("Overall Performance by Sentiment:")
print(sentiment_perf)
print(
    "\nBest-performing sentiment:", sentiment_perf.iloc[0]["sentiment"]
)

# Behavior of Negative and Neutral sentiment posts by Post Type
neg_neu = df[df["sentiment"].str.lower().isin(["negative", "neutral"])]
print("\nNegative & Neutral Posts Breakdown by Post Type:")
print(
    neg_neu.groupby(["sentiment", "post_type"], as_index=False).agg(
        count=("post_type", "count"),
        avg_engagement_rate=("engagement_rate", "mean"),
        avg_impressions=("impression_count", "mean"),
        avg_likes=("likes", "mean"),
    )
)

# Behavior of Negative and Neutral sentiment posts by Category
print("\nNegative & Neutral Posts Breakdown by Category:")
print(
    neg_neu.groupby(["sentiment", "post_category"], as_index=False).agg(
        count=("post_category", "count"),
        avg_engagement_rate=("engagement_rate", "mean"),
        avg_impressions=("impression_count", "mean"),
        avg_likes=("likes", "mean"),
    )
)
```

```python
import pandas as pd

# 1. Content Performance
pt = df.groupby("post_type")["engagement_rate"].mean().sort_values(ascending=False)
pt_score = df.groupby("post_type")["engagement_score"].mean()
top_pt = pt.index[0]

cat = df.groupby("post_category")["engagement_rate"].mean().sort_values(ascending=False)
top_cat = cat.index[0]
top_cat_count = (df["post_category"] == top_cat).sum()

# Optional Country column check
country_col = next((c for c in df.columns if "country" in c.lower() or "region" in c.lower()), None)
if country_col:
    cntry = df.groupby(country_col)["engagement_rate"].mean().sort_values(ascending=False)
    country_line = f"- Country: {cntry.index[0]} showed the highest average engagement rate ({cntry.iloc[0]:.4f})."
else:
    country_line = "- Country: Geographic/country data was not included in this dataset."

# 2. User and Behavioral Trends
age_col = next((c for c in df.columns if "age" in c.lower()), None)
ver_col = next((c for c in df.columns if "verif" in c.lower()), None)
user_notes = []
if age_col:
    corr_val = df[age_col].corr(df["engagement_rate"])
    user_notes.append(f"Age shows a {corr_val:.4f} correlation with engagement rate.")
if ver_col:
    ver_means = df.groupby(ver_col)["engagement_rate"].mean()
    user_notes.append("Verified vs unverified average engagement rates: " + ", ".join([f"{k}: {v:.4f}" for k, v in ver_means.items()]) + ".")
user_line = "- Age and Verified Accounts: " + (" ".join(user_notes) if user_notes else "Age and account verification columns were not included in this dataset.")

df["posted_at"] = pd.to_datetime(df["posted_at"])
df["hour"] = df["posted_at"].dt.hour
def get_tod(h):
    if 5 <= h < 12: return "Morning"
    elif 12 <= h < 17: return "Afternoon"
    elif 17 <= h < 21: return "Evening"
    else: return "Night"
df["time_of_day"] = df["hour"].apply(get_tod)

best_hr = df.groupby("hour")["impression_count"].mean().idxmax()
best_hr_imp = df.groupby("hour")["impression_count"].mean().max()
best_tod = df.groupby("time_of_day")["impression_count"].mean().idxmax()

watch_col = next((c for c in df.columns if "watch" in c.lower() or "duration" in c.lower()), None)
dev_metric = watch_col if watch_col else "engagement_score"
dev = df.groupby("device_type")[dev_metric].mean().sort_values(ascending=False)
top_dev = dev.index[0]

# 3. Sentiment Analysis
sent = df.groupby("sentiment").agg(
    eng_rate=("engagement_rate", "mean"),
    likes=("likes", "mean"),
    impressions=("impression_count", "mean")
).sort_values(by="eng_rate", ascending=False)
top_sent = sent.index[0]

neg_rate = sent.loc[sent.index.str.lower() == "negative", "eng_rate"].values[0] if any(sent.index.str.lower() == "negative") else 0
neg_imp = sent.loc[sent.index.str.lower() == "negative", "impressions"].values[0] if any(sent.index.str.lower() == "negative") else 0
neu_rate = sent.loc[sent.index.str.lower() == "neutral", "eng_rate"].values[0] if any(sent.index.str.lower() == "neutral") else 0
neu_imp = sent.loc[sent.index.str.lower() == "neutral", "impressions"].values[0] if any(sent.index.str.lower() == "neutral") else 0

print(f"""Summary and Key Findings

From the data analysis and the charts created with Matplotlib, Seaborn, and Plotly, here are the main results:

1. Content Performance
- Post Type: {top_pt} had the highest average engagement rate ({pt.iloc[0]:.4f}) and an average engagement score of {pt_score[top_pt]:.2f}.
- Content Category: {top_cat} performed the best among all categories with an average engagement rate of {cat.iloc[0]:.4f} across {top_cat_count} posts.
{country_line}

2. User and Behavioral Trends
{user_line}
- Best Time for Impressions: Posts uploaded during the {best_tod} (especially around {best_hr}:00) got the highest average impressions ({best_hr_imp:,.2f}).
- Device Type Impact: {top_dev} users had the highest average {dev_metric} ({dev.iloc[0]:.2f}).

3. Sentiment Analysis
- Best Sentiment: {top_sent} posts gave the best results overall, with an average engagement rate of {sent.loc[top_sent, 'eng_rate']:.4f} and {sent.loc[top_sent, 'likes']:,.2f} average likes.
- Negative and Neutral Posts: Negative posts had an average engagement rate of {neg_rate:.4f} ({neg_imp:,.2f} average impressions), while neutral posts had an average engagement rate of {neu_rate:.4f} ({neu_imp:,.2f} average impressions).

Overall Takeaway
Posting {top_pt} content under the {top_cat} category around {best_hr}:00 ({best_tod}) gives the best reach and engagement.""")
```
